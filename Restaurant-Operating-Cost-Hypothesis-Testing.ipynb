```python
import pandas as pd
from scipy.stats import norm
from scipy.stats import t
import math
```

# ESTIMATION AND CONFIDENCE INTERVALS

```python
data=[1.13, 1.55, 1.43, 0.92, 1.25,
      1.36, 1.32, 0.85, 1.07, 1.48,
      1.20, 1.33, 1.18, 1.22, 1.29]
```

```python
df=pd.DataFrame(data, columns=['Durability'])
df
```

```python
mean=df['Durability'].mean()
mean
```

```python
std_dev=df['Durability'].std()
std_dev
```

```python
n=len(data)
print(n)
```

```python
df_new=n-1
print(df_new)
```

```python
confidence = 0.99
```

a. Build 99% Confidence Interval Using Sample Standard Deviation

```python
t_critical=t.ppf(1-((1-confidence)/2),df_new)
print(t_critical)
```

```python
margin_of_error = t_critical * (std_dev / math.sqrt(n))
print(margin_of_error)
```

```python
lower = mean - margin_of_error
upper = mean + margin_of_error

print("99% Confidence Interval:")
print('Lower Limit =',round(lower,4))
print("Upper Limit =",round(upper,4))
```

b. Build 99% Confidence Interval Using Known Population Standard Deviation

```python
z_critical = norm.ppf(1 - (1 - confidence) / 2)
print(z_critical)
```

```python
sigma=0.2
```

```python
margin_of_error1 = z_critical * (sigma / math.sqrt(n))
print(margin_of_error1)
```

```python
lower = mean - margin_of_error1
upper = mean + margin_of_error1

print("99% Confidence Interval:")
print("Lower Limit =",round(lower,4))
print("Upper Limit =",round(upper,4))
```

# HYPOTHESIS TESTING

 1. State the Hypotheses statement:
          
*   Null Hypothesis
      * The mean weekly operating cost is equal to the theoretical cost predicted by the model.


*   Alternative Hypothesis
      *   The mean weekly operating cost is greater than the theoretical cost
          predicted by the model.





2. Calculate the Test Statistic

```python
#Given data
sample_mean = 3050
x = 600
n = 25
```

```python
#Theoretical mean
mu = 1000 + (5 * x)
print(mu)
```

```python
#Population standard deviation
sigma = 5 * 25
print(sigma)
```

```python
z = (sample_mean - mu) / (sigma / math.sqrt(n))
print(z)
```

3.Determine the Probability and compare

```python
#Critical value
alpha = 0.05
critical_value = norm.ppf(1 - alpha)
print(critical_value)
```

4. Make a Decision

```python
#Decision
if z > critical_value:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

# 5. Conclusion
There is insufficient evidence to support the restaurant owners' claim that the weekly operating costs are higher than those predicted by the theoretical cost model. Therefore, the existing cost model cannot be rejected based on the given sample data.
