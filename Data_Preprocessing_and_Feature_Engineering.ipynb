Standard Scaling


*   Standard Scaling is preferred for algorithms that depend on the distribution and variance of the data
*   Standard Scaling is preferred when the numerical features have different scales and the machine learning algorithm assumes that the data is approximately normally distributed.
*   It transforms the data to have a mean of 0 and a standard deviation of 1.

Min-Max Scaling


*   Min-Max Scaling is preferred when the data needs to be transformed into a fixed range, typically 0 to 1.
*   It preserves the relationships between the original values while ensuring that all features have the same range.
*   Min-Max Scaling is more sensitive to outliers because the minimum and maximum values determine the scaling range.

Label Encoding

````python
#Label Encoding for variables with more than 5 categories
cols = ['workclass',
        'education',
        'marital_status',
        'occupation',
        'relationship',
        'native_country']
````

````python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df_label = df.copy()
for col in cols:
    df_label[col] = le.fit_transform(df_label[col])
df_label.head()
````

Pros and Cons of One-Hot Encoding


```
Pros:
*   Does not introduce any order among categories.
*   Suitable for nominal categorical variables.
*   Improves performance of many machine learning algorithms.

Cons:
*   Creates many new columns when there are many categories.
*   Increases memory usage.
*   Can make the dataset sparse and slower to process.

```
Pros and Cons of Label Encoding


```
Pros:


*   Simple and easy to implement.
*   Uses less memory since only one column is created.
*   Suitable for ordinal variables and tree-based models.


Cons:


*   Introduces an artificial order among categories.
*   May negatively affect algorithms like Logistic Regression, KNN, and SVM if used on nominal variables.

```





# 3. Feature Engineering:

Create  2 new features

````python
df['total_capital'] = df['capital_gain'] - df['capital_loss']
df['total_capital'].head()
````

Rationale

*   Combines capital gain and capital loss into a single measure.
*   Represents an individual's overall capital income.



````python
df['work_intensity'] = df['hours_per_week'] / 40
df['work_intensity'].head()
````

Rationale


*   A standard full-time work week is approximately 40 hours.
*   This feature may help capture the relationship between working hours and income.



Apply Log Transformation for capital_gain


````python
df['capital_gain_log'] = np.log1p(df['capital_gain'])
````

````python
df['capital_gain_log'].head()
````

The capital_gain feature is highly right-skewed because most values are 0 while a few are extremely large.Log transformation reduces the effect of extremely large values.It makes the distribution more balanced.
