6.	Given a 'marks' variable, print the grade

```python
mark=int(input("Enter your mark: "))
if mark>=90:
  print('Grade : A')
elif mark>=75 and mark<=89:
  print('Grade : B')
elif mark>=50 and mark<=74:
  print('Grade : C')
elif mark<50:
  print('Fail')
else :
  print('Invalid mark')
```

7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
num=int(input("Enter a number:"))
if num>0:
  if num%2==0:
    print(num,' is positive and even.')
  else:
    print(num,' is positive but not even.')
elif num==0:
  print(num,' is zero.')
else:
  print(num,' is negative.')
```

```python
import pandas as pd
data={
    'Name':['Diya','Manu','Laya','Arun','Jith','Veena','Isha','Fidel','Hina','Diya'],
    'Age':[22,34,32,11,23,45,36,34,18,22],
    'City':['Delhi', 'Mumbai', 'Chennai', 'Kolkata', 'Pune',
             'Hyderabad', 'Bangalore', 'Delhi', 'Mumbai', 'Delhi'],
    'Salary':[35000, 45000, None, 50000, 32000,
               60000, None, 48000, 42000, 35000]
}
df1=pd.DataFrame(data)
df1
```

```python
df1.info()
```

```python
df1.describe()
```

```python
#Select 'Name' and 'City' columns
df1[['Name','City']]
```

```python
#Drop 'City' column
df2=df1.drop(['City'],axis=1)
df2
```

```python
#Fill any missing values in 'Salary' column with the mean
df1['Salary'] = df1['Salary'].fillna(df1['Salary'].mean())
df1
#Salary of Laya and isha filled with mean salary
```

```python
#Remove any duplicate rows
df1.drop_duplicates(inplace=True)
df1
#Removed row 0 and 9
```
